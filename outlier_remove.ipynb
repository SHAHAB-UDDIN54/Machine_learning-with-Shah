```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
sns.boxplot(x="sqft_living",data =dataset)
plt.show()
```

```python
q1 = dataset["sqft_living"].quantile(0.25)
q2 = dataset["sqft_living"].quantile(0.75)

IQR = q2-q1
```

```python
lower_limit = q1-1.5*IQR

```

```python
upper_limit = q2+1.5*IQR
```

```python
df_clean = dataset[(dataset["sqft_living"] >= lower_limit) & (dataset["sqft_living"] <= upper_limit)]
print(df_clean)
```
