```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv("/content/house price.csv")
dataset.head()
```

```python
dataset["bedrooms"].fillna(dataset["bedrooms"].mean(),inplace=True)
```

```python
sns.distplot(dataset["bedrooms"])
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
ms = MinMaxScaler()
```

```python
ms.fit(dataset["bedrooms"].values.reshape(-1, 1))
```

```python
dataset["bedrooms"]=ms.fit_transform(dataset["bedrooms"].values.reshape(-1, 1))
```

```python
dataset.head(3)
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title("Before")
sns.distplot(dataset["bedrooms"])

plt.subplot(1,2,2)
plt.title("After")
sns.displot(dataset["bedrooms"])
plt.show()
```
